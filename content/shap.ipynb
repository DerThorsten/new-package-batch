# SHAP: explain a machine-learning model in your browser

[SHAP](https://shap.readthedocs.io) explains the predictions of a model with Shapley values: how much did each feature push a prediction up or down? Here we train a random forest on scikit-learn's diabetes dataset (it ships with scikit-learn, so nothing is downloaded) and look at it with SHAP's plots.

```python
import matplotlib.pyplot as plt
import shap
from sklearn.datasets import load_diabetes
from sklearn.ensemble import RandomForestRegressor

X, y = load_diabetes(return_X_y=True, as_frame=True)
X.head()
```

The target is a measure of disease progression one year after baseline. Features include age, sex, body mass index (`bmi`), blood pressure (`bp`) and six blood serum measurements (`s1` ... `s6`).

```python
model = RandomForestRegressor(n_estimators=100, max_depth=6, random_state=0).fit(X, y)

explainer = shap.TreeExplainer(model)
shap_values = explainer(X)
shap_values.shape
```

## Which features matter most?

Each dot is one patient. Red means a high feature value, blue a low one; the x position shows how much that feature changed the prediction.

```python
shap.plots.beeswarm(shap_values)
```

```python
shap.plots.bar(shap_values)
```

## Why did the model predict this for one patient?

The waterfall plot starts at the average prediction and adds the contribution of each feature.

```python
shap.plots.waterfall(shap_values[0])
```

```python
shap.plots.force(shap_values[0], matplotlib=True)
```

## How does one feature act?

The dependence plot shows the effect of body mass index on the prediction, colored by the feature it interacts with most.

```python
shap.plots.scatter(shap_values[:, "bmi"], color=shap_values)
```
